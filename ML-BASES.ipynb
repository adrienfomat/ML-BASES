## Variables et Fonctions

```python
def e_potentielle(masse, hauteur, e_limite, g=9.81):
    E = masse * hauteur * g
    if E > e_limite:
        print("L'energie potentielle: ",E, "Joules est superieur a l'energie limite " ,e_limite)
    else:
        print("L'energie potentielle: ",E, "Joules est inferieur a l'energie limite " ,e_limite)
    return E

resultat = e_potentielle(masse= 120, hauteur = 3, e_limite= 3200, g=10)
print("l'energie potentielle est : ", resultat)
```

## LES STRUTURES DE CONTROLES : IF/ELSE, WHILE, FOR 

```python
#Suite de Fibonacci 
def fibonacci(n):
    res = []
    if n == 0 :
        res.append(0)
    elif n == 1 :
        res.append(1)
        
    a,b = 0, 1
    i = 2
    while i <= n:
        a , b = b, a + b
        i +=1
        res.append(b)
    return res
    
n =10
res = fibonacci(n)
print ("La suite des nombres de Fibonacci de", n , "est :", res)
```

```python
#fibonacci qui imprime toute les valeurs avant n 
#Suite de Fibonacci 
def fibonacci(n):
    a = 0
    b =  1
    sequence = [a]
    while b < n:
        a , b = b, a + b
        sequence.append(a)
    return sequence
    
n =10
res = fibonacci(n)
print ("La suite des nombres de Fibonacci de", n , "est :", res)
```

```python
#Suite de Fibonacci 
def fibonacci(n):
    if n < 0 :
        return []
    elif n ==0 :
        return [0]
    elif n == 1 :
        return [0,1]
        
    sequence = [0,1]
    i =2
    while i <= n:
        next_val = sequence[i-2] + sequence[i-1]
        sequence.append(next_val)
        i +=1
    return sequence
    
n =10
res = fibonacci(n)
print ("La suite des nombres de Fibonacci de", n , "est :", res)
```

# LES DICTIONNAIRES

```python
classeur = {
    "positif": [],
    "negatif": [],
}
```

```python
def trier(classeur, nombre):
    if nombre < 0:
        classeur["negatif"].append(nombre)
    else:
        classeur["positif"].append(nombre)
    return classeur
```

```python
nombre = [ 1, -2, 0, 16, -15, -9, 10, 8]
for i in nombre:
    show = trier(classeur, i)
show
```

## List comprehension 

```python
# Fonction qui renvoie tout le carré de tout mes nombres entres 0 et 10
liste_1 = [] 
for i in range (0, 11, 1):
    liste_1.append(i**2)
liste_1
```

```python
#Avec les list comprehension 
liste_2 = [i**2 for i in range (0,11,1)]
liste_2
```
