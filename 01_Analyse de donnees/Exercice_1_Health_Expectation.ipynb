## 4.1 - life expentency

LIFE_EXPECTANCY represente l'esperance de vie des individus d'un pays en une année donnée, c'est-à-dire le nombre moyen d'années qu'un individu peut s'attendre à vivre à partir de sa naissance.

La Moyenne donnée par describe represente donc la durée moyenne de vie des indivucus de maniere generale dans notre dataset tout pays et années confondus.

## 6. Oups... Comment améliorer ce graphique ?
