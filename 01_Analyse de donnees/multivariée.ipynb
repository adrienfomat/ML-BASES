```python
import pandas as pd
import matplotlib 
import matplotlib.pyplot as plt
import seaborn as sns

print("Pandas version:", pd.__version__)
print("Matplotlib version:", matplotlib.__version__)
print("Seaborn version:", sns.__version__)
```

# Data

```python
df = sns.load_dataset('tips')

df.head()
```

# Analyse Multivariée

## Discret/ Discret

```python
pd.crosstab(df['size'], df['day'], normalize=False)
```

```python
sns.heatmap(pd.crosstab(df['size'], df['day']), annot=True)
```

## Discret/ Continue

```python
df.groupby("sex")["tip"].mean()
```

```python
df.groupby("sex")["tip"].median()
```

```python
df.groupby("sex")["tip"].describe()
```

```python
df.groupby("sex")["tip"].mean().plot(kind='bar')
plt.show()
```

```python
sns.catplot(data=df, x="sex", y="tip",kind="bar")
```

```python
sns.displot(data=df, x="tip", hue="sex")
```

```python
sns.catplot(data=df, x="sex", y="tip",kind="box")
```
